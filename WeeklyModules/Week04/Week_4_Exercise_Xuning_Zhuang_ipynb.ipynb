```python
# 2.d
ring_locations = ['Bolivia', 'Chile', 'Ecuador', 'Peru', 'Costa Rica', 'Guatemala', 'Mexico', 'Japan', 'Philippines',
                  'Indonesia', 'New Zealand', 'Canada', 'Fiji', 'Alaska', 'Washington', 'California', 'Russia',
                  'Taiwan', 'Tonga', 'Kermadec Islands']
```

```python
df['parsed_place'].isin(ring_locations)
```

```python
df['ring_of_fire'] = (df['parsed_place'].isin(ring_locations) | df['parsed_place'].str.contains('Antarctic', case = False, na = False))
print(df['ring_of_fire'])
```

```python
# 2.e
ring_of_fire_count = df['ring_of_fire'].sum()
ring_outside_of_fire_count = (~df['ring_of_fire']).sum()

print(ring_of_fire_count)
print(ring_outside_of_fire_count)
```

```python
# 2.f
ring_tsunami = df.loc[df['ring_of_fire'], 'tsunami'].sum()

print(ring_tsunami)
```
