```python
#f
stocks = pd.read_csv('faanf.csv')

stocks['date'] = pd.to_datetime(stocks['date'])
stocks['volume'] = stocks['volume'].astype(int)

stocks = stocks.sort_values(['date', 'ticker'])
```
