```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/MSSP6070/data/InternetSales.csv', encoding='latin-1')
df['OrderDate'] = pd.to_datetime(df['OrderDate'])
```

```python
yearly_sales = df.groupby(df['OrderDate'].dt.year)['SalesAmount'].sum()

print(yearly_sales)
```

```python
yearly_sales.plot(kind='bar')

plt.xlabel('Year')
plt.ylabel('Sales Amount')
plt.title('Yearly Sales')
plt.xticks(rotation=45)

plt.savefig('InternetSales_Yearly_Sales.pdf', format='pdf', bbox_inches='tight')
plt.show()
```
