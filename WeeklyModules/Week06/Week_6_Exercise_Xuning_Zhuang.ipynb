```python
import pandas as pd

df = pd.read_csv('/content/MSSP6070/data/Student Performance Data.csv')
df.head()
```

```python
#for culculating more easily, I choose five columns, which are 'gender', 'math score', 'reading score', 'writing score'and 'total score'
score_columns = ['gender', 'math score', 'reading score', 'writing score', 'total score']
```

```python
# In my understanding, lambda menas define a new function.
# so that for the name of the new function I defined is 'student_scores', and the name of 'data' is the parameter, the expression after : is the function that I want to executive
# so that in the new function I creat, the 'student_scores' is the name of my new function, 'data' is the parameter, the 'student_scores' function need to accept a dataframe.
# for the copy, means I want to copy all of the data of the 'score_columns' extract from the 'data' dataframe
# and finally the data of the 'score_columns' will be returned to a new dataframe
student_scores = lambda data: data[score_columns].copy()
result = student_scores(df) # the original one should be result = student_scores(data), so that now df=data, df is the parameter
print(result.head())
```
