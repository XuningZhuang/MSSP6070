```python
import os
import time
import requests

repo_folder = '/content/MSSP6070'

download_folder = os.path.join(
    repo_folder,
    'WeeklyModules',
    'Week09',
    'images'
)

os.makedirs(download_folder, exist_ok=True)

print('Images will be saved in:', download_folder)
```

```python
search_term = 'Philadelphia'

url = 'https://www.flickr.com/services/feeds/photos_public.gne'

params = {'tags': search_term, 'tagmode': 'all', 'format': 'json', 'nojsoncallback': 1}

headers = {'User-Agent': 'StudentImageDownloader/1.0 (educational project)'}

response = requests.get(url, params=params, headers=headers, timeout=30)    #Dear professor，the following Chinese text is just some notes I made for myself. .get就是python的一种取值方式的命令

response.raise_for_status()

data = response.json()

print('Search term:', search_term)
print('Number of images found:', len(data['items']))
```

```python
downloaded = 0

for item in data['items']:

    image_url = item['media']['m']

    filename = 'philadelphia_' + str(downloaded + 1) + '.jpg'

    save_path = os.path.join(download_folder, filename)

    try:
        print('Downloading:', image_url)

        image_response = requests.get(image_url, headers=headers, timeout=30)   #发送请求时，附带前面定义好的 HTTP headers。Headers 可以包含程序的标识信息、接受的数据类型等。

        image_response.raise_for_status()   #检查状态

        with open(save_path, 'wb') as image_file:    #wb means white binary，意思就是以二进制写入模式创建文件，因为图片是二进制数据，不是普通文本

            for chunk in image_response.iter_content(100000):   #chunk就把服务器返回的图片数据分成一块一块，每次处理大约 100,000 bytes。把当前这一块图片数据写入刚刚创建的文件
                image_file.write(chunk)

        downloaded += 1

        print('Saved:', filename)

    except requests.exceptions.RequestException as e:

        print('Could not download this image:', e)

    time.sleep(1)   #Wait before requesting the next image，让程序等待 1 秒，再继续执行

print()
print('Finished.')
print('Number of images downloaded:', downloaded)
```

```python
downloaded_files = os.listdir(download_folder)

print('Number of files:', len(downloaded_files))

for filename in downloaded_files:
    print(filename)
```
