```python
import requests

headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)'}           # 定义 HTTP headers

api_url = "https://commons.wikimedia.org/w/api.php"                     # 向 Wikimedia 请求关于 Philadelphia 的图片信息
params = {"action": "query", "format": "json", "generator": "search",
    "gsrsearch": "Philadelphia", "gsrnamespace": "6", "gsrlimit": "10",      # 拿10张照片
    "prop": "imageinfo", "iiprop": "url", "iiurlwidth": "1280"}

print("Fetching data...")
response = requests.get(api_url, headers=headers, params=params)
data = response.json()

if 'query' in data and 'pages' in data['query']:
    pages = data['query']['pages']
    print(f"Successfully fetched {len(pages)} pages.")
else:
    pages = {}
    print("Failed to fetch data.")
```

```python
import time

downloaded = 0

for page in pages.values():
    if 'imageinfo' not in page:        # 检查确保有 imageinfo
        continue

    image_info = page['imageinfo'][0]
    image_url = image_info.get('thumburl', image_info.get('url'))

    if not image_url:
        continue

    try:
        print('Downloading:', image_url)
        image_response = requests.get(image_url, headers=headers)


        if image_response.status_code == requests.codes.ok:
            downloaded += 1
            filename = f'philadelphia_{downloaded}.jpg'
            save_path = os.path.join('week9_images', filename)

            file = open(save_path, 'wb')
            file.write(image_response.content)
            file.close()
            print('Saved:', filename)
        else:
            print('Could not download, status code:', image_response.status_code)

    except Exception as e:
        print('Error:', e)

    time.sleep(1)

print(f"\nFinished. Number of images downloaded: {downloaded}")
```
